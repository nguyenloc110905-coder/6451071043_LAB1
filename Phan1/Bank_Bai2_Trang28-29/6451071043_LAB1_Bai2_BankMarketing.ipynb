```python
from pathlib import Path
import os, re, io, zipfile, time, warnings, string
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules
from sklearn.preprocessing import MinMaxScaler
from bs4 import BeautifulSoup
warnings.filterwarnings('ignore', category=FutureWarning)
sns.set_theme(style='whitegrid')
HO_TEN = 'NGUYỄN LỘC'
MSSV = '6451071043'
BASE = Path('/content/lab1_data') if Path('/content').exists() else Path('/mnt/data')
BASE.mkdir(parents=True, exist_ok=True)
OUTPUT = BASE / 'results'; OUTPUT.mkdir(exist_ok=True)
def identity(): print(f'Họ tên: {HO_TEN} | MSSV: {MSSV}')
def pick_file(keywords, extensions=('.csv','.xlsx','.xls','.txt','.html','.htm')):
    fs=[p for p in BASE.rglob('*') if p.is_file() and p.suffix.lower() in extensions and not OUTPUT in p.parents]
    for keys in keywords:
        matches=[f for f in fs if all(w in f.name.lower() for w in keys)]
        if matches:return sorted(matches,key=lambda f:len(f.name))[0]
    raise FileNotFoundError(f'Không tìm thấy tập tin chứa từ khóa {keywords} trong {BASE}. File hiện có: {[x.name for x in fs[:30]]}')
def read_table(file, **kw):
    if file.suffix.lower() in ('.xls','.xlsx'):return pd.read_excel(file,**kw)
    for enc in ['utf-8-sig','utf-8','cp1252','latin-1']:
        try:
            df=pd.read_csv(file,encoding=enc,**kw)
            if df.shape[1]==1 and ';' in str(df.columns[0]):df=pd.read_csv(file,encoding=enc,sep=';',**kw)
            return df
        except UnicodeDecodeError:continue
    return pd.read_csv(file,encoding='latin-1',**kw)
def norm_cols(df):
    df=df.copy(); df.columns=[re.sub(r'[^a-z0-9]+','',str(c).strip().lower()) for c in df.columns]
    return df
def rule_table(freq, confidence):
    if freq.empty:return pd.DataFrame()
    try:r=association_rules(freq,metric='confidence',min_threshold=confidence,num_itemsets=int(freq.attrs.get('n_transactions',0)) or None)
    except TypeError:r=association_rules(freq,metric='confidence',min_threshold=confidence)
    if len(r):
        r=r.copy()
        for c in ['antecedents','consequents']:r[c]=r[c].map(lambda s:', '.join(sorted(map(str,s))))
    return r
identity(); print('Thư mục dữ liệu:', BASE)
# Tự dò dữ liệu được Colab upload vào /content hoặc LAB1_data.
def ensure_dataset(filename):
    for d in [BASE,Path('/content'),Path('/content/LAB1_data'),Path('/mnt/data')]:
        p=d/filename
        if p.exists(): return p
    try:
        from google.colab import files
        print('Upload dataset:',filename)
        upl=files.upload()
        for name,b in upl.items(): (BASE/Path(name).name).write_bytes(b)
        if (BASE/filename).exists():return BASE/filename
    except ImportError:pass
    raise FileNotFoundError(filename)
```

### Kết quả bài làm

```python
identity()
bank_file = ensure_dataset('bank-full.csv')
bank=norm_cols(read_table(bank_file));print('File:',bank_file.name,'|',bank.shape)
display(bank.head());bank.info();display(bank.describe(include='all').T)
print('Missing values (kể cả chuỗi unknown):')
display((bank.isna().sum()+bank.astype(str).apply(lambda s:s.str.lower().eq('unknown').sum())).sort_values(ascending=False).head(15).to_frame('Số thiếu/unknown'))
identity()
bank_clean=bank.copy().replace(r'^\s*$',np.nan,regex=True).replace(['unknown','Unknown','UNKNOWN'],np.nan)
for c in bank_clean.columns:
    if pd.api.types.is_numeric_dtype(bank_clean[c]):bank_clean[c]=bank_clean[c].fillna(bank_clean[c].median())
    else:
        mode=bank_clean[c].mode(dropna=True)
        bank_clean[c]=bank_clean[c].fillna(mode.iloc[0] if len(mode) else 'Missing')
num=bank_clean.select_dtypes(include='number').columns.tolist()
if num:
    scaled=MinMaxScaler().fit_transform(bank_clean[num]);bank_scaled=pd.DataFrame(scaled,columns=[c+'_minmax' for c in num],index=bank_clean.index)
    display(bank_scaled.head())
cat=[c for c in bank_clean.columns if c not in num]
bank_encoded=pd.get_dummies(bank_clean,columns=cat,dtype=int)
print('Sau mã hóa:',bank_encoded.shape);display(bank_encoded.head())
if 'y' in bank_clean.columns:
    print('Phân bố khách hàng đăng ký tiền gửi:')
    display(bank_clean['y'].value_counts(dropna=False))

    if 'job' in bank_clean.columns:
        print('Tỷ lệ đăng ký theo nghề nghiệp:')
        display(
            pd.crosstab(
                bank_clean['job'],
                bank_clean['y'],
                normalize='index'
            ).round(3)
        )
bank_encoded.to_csv(OUTPUT/'BankMarketing_encoded.csv',index=False);print('Đã lưu dữ liệu đã mã hóa.')
```
